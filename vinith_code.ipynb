```python
import pandas as pd
df = pd.read_csv("41598_2021_87762_MOESM1_ESM.csv")
```

```python
print(list(df.columns))
```

```python
df_sorted = df.sort_values(by="TMC-Shapley Value", ascending=False)
df_sorted[200:210]
```

```python
top100 = df_sorted[:100]
bottom100 = df_sorted[200:]
len(df_sorted), len(top100)
```

```python
print("No. of Pneumonia Data in Most Valuable:", top100["Pneumonia (ChestX-ray14)"].sum())
print("No. of Pneumonia Data in Least Valuable:", bottom100["Pneumonia (ChestX-ray14)"].sum())
```

```python
count1 = ((top100["Pneumonia (Radiologists Final)"]==0) & (top100["Pneumonia (ChestX-ray14)"]==1)).sum()
count2 = ((top100["Pneumonia (Radiologists Final)"]==1) & (top100["Pneumonia (ChestX-ray14)"]==0)).sum()
print("No of Mislabelled data as Pneumonia in Most Valuable:", count1)
print("No of Mislabelled data as not Pneumonia in Most Valuable:", count2)
```

```python
count1 = ((bottom100["Pneumonia (Radiologists Final)"]==0) & (bottom100["Pneumonia (ChestX-ray14)"]==1)).sum()
count2 = ((bottom100["Pneumonia (Radiologists Final)"]==1) & (bottom100["Pneumonia (ChestX-ray14)"]==0)).sum()
print("No of Mislabelled data as Pneumonia in Least Valuable:", count1)
print("No of Mislabelled data as not Pneumonia in Least Valuable:", count2)
```
